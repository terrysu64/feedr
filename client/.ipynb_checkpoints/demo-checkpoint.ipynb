# Feedr — Prediction Market Data Explorer

This notebook loads data produced by the feedr pipeline and does basic analysis on prediction market price history.

## 1. Setup

First, run the pipeline if you haven't already:
```bash
bazel run //experimental/terry-su_data/feedr:feedr -- --output-dir /tmp/feedr_output --verbose
```

````python
import json
import os
from datetime import datetime, timezone

import matplotlib.pyplot as plt
import pandas as pd
import pyarrow.parquet as pq

OUTPUT_DIR = "/Users/terrysu64/Desktop/feedr stuff/feedr_output"

# Verify output exists
assert os.path.exists(os.path.join(OUTPUT_DIR, "manifest.json")), (
    "No output found. Run the pipeline first."
)
print("Output directory ready.")
````

## 2. Discover Available Markets

````python
with open(os.path.join(OUTPUT_DIR, "manifest.json")) as f:
    manifest = json.load(f)

markets = manifest["markets"]
print(f"Last updated: {manifest['last_updated']}")
print(f"Total markets: {len(markets)}\n")

# Show as a table
summary = pd.DataFrame([
    {
        "question": m["question"][:60],
        "volume": f"${m['volume_usd']:,.0f}",
        "prices": m["num_prices"],
        "resolution": m.get("resolution", ""),
        "path": m["path"],
    }
    for m in sorted(markets, key=lambda x: x["volume_usd"], reverse=True)
])
summary
````

## 3. Load & Plot a Single Market

````python
# Pick the highest-volume market
market = sorted(markets, key=lambda x: x["volume_usd"], reverse=True)[0]
print(f"Market: {market['question']}")
print(f"Volume: ${market['volume_usd']:,.0f}")
print(f"Resolution: {market.get('resolution', 'N/A')}")

# Load price data
prices_path = os.path.join(OUTPUT_DIR, market["path"], "prices.parquet")
df = pq.read_table(prices_path).to_pandas()
df["date"] = pd.to_datetime(df["timestamp"], unit="s")
df = df.sort_values("timestamp")

print(f"\nPrice observations: {len(df)}")
print(f"Date range: {df['date'].min().date()} to {df['date'].max().date()}")
print(f"Outcomes: {list(df['outcome_label'].unique())}")
df.head()
````

````python
# Plot price timeseries for each outcome
fig, ax = plt.subplots(figsize=(12, 5))

for label in df["outcome_label"].unique():
    subset = df[df["outcome_label"] == label]
    ax.plot(subset["date"], subset["price"], label=label, linewidth=1.5)

ax.set_title(market["question"], fontsize=13)
ax.set_ylabel("Price (probability)")
ax.set_ylim(-0.05, 1.05)
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 4. Multi-Market Comparison

Overlay the "Yes" outcome price for all markets to compare trajectories.

````python
fig, ax = plt.subplots(figsize=(12, 5))

for m in sorted(markets, key=lambda x: x["volume_usd"], reverse=True):
    path = os.path.join(OUTPUT_DIR, m["path"], "prices.parquet")
    if not os.path.exists(path):
        continue
    mdf = pq.read_table(path).to_pandas()
    mdf["date"] = pd.to_datetime(mdf["timestamp"], unit="s")
    yes = mdf[mdf["outcome_label"] == "Yes"].sort_values("timestamp")
    if yes.empty:
        continue
    label = m["question"][:50]
    ax.plot(yes["date"], yes["price"], label=label, linewidth=1.5)

ax.set_title("Market Comparison — 'Yes' Outcome Prices", fontsize=13)
ax.set_ylabel("Price (probability)")
ax.set_ylim(-0.05, 1.05)
ax.legend(fontsize=8)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 5. Analysis — Convergence Speed

For markets that resolved to "Yes", how many days did it take to go from 50% to 90% confidence?

````python
convergence_results = []

for m in markets:
    path = os.path.join(OUTPUT_DIR, m["path"], "prices.parquet")
    if not os.path.exists(path):
        continue

    mdf = pq.read_table(path).to_pandas()
    mdf["date"] = pd.to_datetime(mdf["timestamp"], unit="s")

    # Look at the resolved outcome
    resolution = m.get("resolution", "")
    if not resolution:
        continue

    outcome_df = mdf[mdf["outcome_label"] == resolution].sort_values("timestamp")
    if outcome_df.empty:
        continue

    # Find first day above 50% and first day above 90%
    above_50 = outcome_df[outcome_df["price"] >= 0.5]
    above_90 = outcome_df[outcome_df["price"] >= 0.9]

    if above_50.empty or above_90.empty:
        continue

    date_50 = above_50.iloc[0]["date"]
    date_90 = above_90.iloc[0]["date"]
    days_to_converge = (date_90 - date_50).days

    convergence_results.append({
        "question": m["question"][:60],
        "resolution": resolution,
        "first_above_50%": date_50.date(),
        "first_above_90%": date_90.date(),
        "days_50_to_90": days_to_converge,
        "volume_usd": m["volume_usd"],
    })

if convergence_results:
    conv_df = pd.DataFrame(convergence_results)
    display(conv_df)
else:
    print("No resolved markets with convergence data found.")
    print("Try running the pipeline with more markets (increase max_markets_per_run in feedr_config.yaml).")
````
